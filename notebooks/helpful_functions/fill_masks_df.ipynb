# Populate mask file paths into the analysis dataframe

Resolve the on-disk paths of per-image masks and write them into the appropriate columns of the analysis dataframe.

```python
masks_df_path = Path(input())
```

```python
masks_df = pd.read_csv(masks_df_path)
masks_df.head()
```

```python
exp = masks_df['basename'].str.split('_').str[0]
masks_df['UID'] = masks_df['wellID'] + '_' + masks_df['scene'] + '_' + masks_df['ROI']
UID = masks_df.pop('UID')
masks_df.insert(4, 'UID', UID)
masks_df.head()
```

```python
well_cond_df_path = Path(input())
well_cond_df = pd.read_csv(well_cond_df_path)
well_cond_df['wellID'] = well_cond_df['experiment'] + '_' + well_cond_df['wellID']
```

```python
well_cond_df.head()
```

```python
len_df_premerge = len(masks_df)

masks_df = pd.merge(masks_df, well_cond_df, on = ["wellID"], how = "left", suffixes = [None, '_y'], validate='many_to_one')
assert len_df_premerge == len(masks_df)

masks_df.head()
```

```python
utils.safe_save_csv(masks_df, masks_df_path)
```
