# Standardize well conditions: pharmacological-actin experiments

Map condition labels from the pharmacological-actin (LatA, JLY, Jasp) experiment series onto the canonical condition vocabulary used by the manuscript figures.

```python
df_path = Path(input())
```

```python
df = pd.read_csv(df_path)
df.head()
```

```python
df['wellID'] = df['wellID'].str.replace('CE029-', '')
df['wellID'] = df['wellID'].str.replace('CE030-', '')
df['wellID'] = df['wellID'].str.replace('CE031-', '')
```

```python
ch_content_list = ['cell', 'caax']
chs = [0, 1, 2]

for ch_content in ch_content_list:
    
    cond = [(df[f'ch{ch}']==ch_content) for ch in chs]
    vals = chs

    df[f'{ch_content} ch'] = np.select(cond, vals, -1)

df
```
