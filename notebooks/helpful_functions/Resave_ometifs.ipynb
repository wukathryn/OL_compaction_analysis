## Resave OmeTiffFiles

### Use case: sometimes ome-tifs saved in Fiji do not display channels correctly when reopened in Fiji. Here, we re-save the ome-tifs to fix the issue.

```python
from pathlib import Path

from bioio import BioImage
import bioio_ome_tiff
import bioio_tifffile
import bioio_imageio
from bioio.writers import OmeTiffWriter

import sys
src_path = str(Path.cwd().parent.parent)
if src_path not in sys.path:
    sys.path.append(src_path)
import src.d00_utils.utilities as utils
from src.d01_init_proc import align_imgs
```

```python
metadata_imgpath = Path(input())
```

```python
input_dirpath = Path(input())
```

```python
# save metadata where it's been lost
img_file = BioImage(metadata_imgpath, reader=bioio_ome_tiff.Reader)

for imgpath in input_dirpath.glob('*.ome.tif'):
    img = BioImage(imgpath, reader=bioio_tifffile.Reader).data
    print(imgpath.name)
    print(img.shape)
    ome_metadata = utils.construct_ome_metadata(img, img_file)
    OmeTiffWriter.save(img, imgpath, ome_xml = ome_metadata)
print('done!')
```
