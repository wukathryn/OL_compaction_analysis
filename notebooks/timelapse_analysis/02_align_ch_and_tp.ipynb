```python
from pathlib import Path
import pandas as pd
import numpy as np

import microscopy_analysis.d00_utils.utilities as utils
import microscopy_analysis.d00_utils.dirnames as dn
from microscopy_analysis.d01_init_proc import alignment_core

from bioio import BioImage
import bioio_ome_tiff
from bioio.writers import OmeTiffWriter

import matplotlib.pyplot as plt
from tqdm import tqdm
```

```python
input_dirpath = Path(input())
```

```python
# Create output directory for time and channel aligned images
assert input_dirpath.is_dir()
proc_dirpath = utils.get_proc_dirpath(input_dirpath)
output_dirpath = proc_dirpath / 't_and_ch_aligned'
output_dirpath.mkdir(exist_ok=True)
```

```python
# input channel to use a reference for channel alignment
ch_ref = 0

imgpaths = [path for path in input_dirpath.glob('*.ome.tif')]

for imgpath in tqdm(imgpaths):
    
    if not (output_dirpath / imgpath.name).is_file():
        img_file = BioImage(imgpath, open=bioio_ome_tiff.Reader)
        img = img_file.data

        print(imgpath.name)
        
        # align all timepoints and all ch
        aligned_img, tp_shifts = alignment_core.align_timepoints(img, ch_ref=0)
        aligned_img, ch_shifts = alignment_core.align_channels(aligned_img, ch_ref=ch_ref)
        
        ome_metadata = utils.construct_ome_metadata(aligned_img, img_file)
        OmeTiffWriter.save(aligned_img, output_dirpath / imgpath.name, ome_xml=ome_metadata)
    
print('Done!')
```
