```python
df_subset = df[df['condition'] == 'Control']
cond = 'ctrl'
figname = f'mean_caax_int_caax_vs_perc_comp_{cond}'
x_label = 'mean caax int (caax)'
y_label = '% compaction'
sns.scatterplot(df_subset, x=x_label, y=y_label, legend=False, size=3, color='k')
plt.savefig(graphs_dir / f'{figname}.png')
plt.ylim(0, 1)

sns.lmplot(x=x_label, y=y_label, data=df_subset)
plt.show()

sns.lmplot(x=x_label, y=y_label, data=df_subset, hue='experiment')
plt.show()

slope, intercept, r_value, p_value, std_err = stats.linregress(df_subset[x_label],df_subset[y_label])
print(f'{cond}: slope: {slope}, intercept: {intercept}, r_value: {r_value}, p_value: {p_value}, std_err: {std_err}')

figname = f'caax_intden_vs_perc_comp_{cond}'
df_subset.loc[:, 'caax integrated density'] = df_subset['mean caax int (cell)'] * df_subset['cell area']
x_label = 'caax integrated density'
sns.scatterplot(df_subset, x=x_label, y=y_label, legend=False, size=3, color='k')
plt.savefig(graphs_dir / f'{figname}.png')
plt.ylim(0, 1)
plt.show()

df_subset = df[df['condition']=='DeAct']
cond = 'DeAct'
figname = f'mean_caax_int_cell_vs_perc_comp_{cond}'
x_label = 'mean caax int (caax)'
y_label = '% compaction'
sns.scatterplot(df_subset, x=x_label, y=y_label, legend=False, size=3, color='k')
plt.savefig(graphs_dir / f'{figname}.png')
plt.ylim(0, 1)
plt.show()
slope, intercept, r_value, p_value, std_err = stats.linregress(df_subset[x_label],df_subset[y_label])
print(f'{cond}: slope: {slope}, intercept: {intercept}, r_value: {r_value}, p_value: {p_value}, std_err: {std_err}')

sns.lmplot(x=x_label, y=y_label, data=df_subset)
plt.show()

figname = f'GFPcaax_intden_vs_perc_comp_{cond}'
df_subset.loc[:, 'caax integrated density'] = df_subset['mean caax int (cell)'] * df_subset['cell area']
x_label = 'caax integrated density'
sns.scatterplot(df_subset, x=x_label, y=y_label, legend=False, size=3, color='k')
plt.savefig(graphs_dir / f'{figname}.png')
plt.ylim(0, 1)
plt.show()
```

```python
smallpts_fillcolor = '#DBDBDB'
smallpts_edgecolor = '#AFAFAF'
ticks_fontsize = 8
axislabel_fontsize = 10
linewidth = 0.5

rc = {'svg.fonttype':'none', 
      'font.family':'Arial',
      'font.size':1,
      'figure.figsize': (4.5,5),
      'figure.dpi': 150,
      'axes.linewidth': 0.5,
      'axes.labelweight':'bold',
      'axes.labelsize': axislabel_fontsize,
      'axes.labelpad':7.5
     }

sns.set(rc)
sns.set_style("ticks")

x_order = ['Control', 'DeAct']
x = 'condition'

actincols = [y for y in ycols if 'actin' in y]
for y in actincols:
    
    ctrl = biorep_avg_df.loc[biorep_avg_df['condition']==x_order[0], y]
    deact = biorep_avg_df.loc[biorep_avg_df['condition']==x_order[1], y]
    stat, pvalue = stats.ttest_rel(ctrl, deact)

    graphname = y.replace('(', '').replace(')', '').replace(' ', '_').replace('%', 'perc') + '.svg'
    fig, ax = plt.subplots(figsize=(2.75, 3), dpi=150)
    sns.swarmplot(x=x, y=y, data = df, size=3, ax=ax, color=smallpts_fillcolor, edgecolor=smallpts_edgecolor, linewidth=0.2)
    sns.swarmplot(x=x, y=y, data = actin_df, size=3, ax=ax, hue='experiment', edgecolor=smallpts_edgecolor, linewidth=0.2)
    sns.swarmplot(x=x, y=y, order=x_order, data = biorep_avg_df, hue='condition', legend=False, size=8)
    sns.pointplot(data=biorep_avg_df, x=x, y=y, linestyle='', errorbar='se', marker='_', markersize=30, markeredgewidth=3, zorder=3, color='k', capsize=0.1)
    fig.suptitle(f't={np.round(stat, 3)}, p-val={np.round(pvalue, 4)}', fontsize=12)
    sns.despine()
    plt.ylim(0)
    ax.xaxis.label.set_visible(False)
    ax.tick_params(direction='out', width=linewidth, labelsize=ticks_fontsize)
    plt.savefig(graphs_dir / graphname, format='svg', bbox_inches='tight')
    plt.show()
    
```
